Use Local Console mode

Read

```
# We specify the host and the port as options
from pyspark.sql.functions import *
df = spark.readStream.format("socket").option("host", "localhost").option("port", 9999).load()

```

Transform

```
w_df= df.select(expr("explode(split(value,' ')) as word"))
df_count =w_df.groupBy("word").count()
spark.conf.set("spark.sql.shuffle.partitions", 3)
```

Write -- Complete mode

```
query = df_count \
    .writeStream \
    .outputMode("complete") \
    .format("console") \
    .start()

```

Write -- Update mode

```
query = df_count \
    .writeStream \
    .outputMode("update") \
    .format("console") \
    .start()
```

Write - Append mode

```
query = df_count \
    .writeStream \
    .outputMode("append") \
    .format("console") \
    .start()
```
